# 형변환 (Casting)

- 형변환은 값을 다른 자료형으로 바꾸는 작업이다.
- Python에서는 상황에 따라 자동으로 바뀌는 **암묵적 형변환**과, 개발자가 직접 함수로 바꾸는 **명시적 형변환**이 있다.
- 조건문에서는 값 자체가 `True` 또는 `False`처럼 평가되기도 한다.

## 1. 암묵적 형변환

암묵적 형변환은 Python이 연산 과정에서 자동으로 자료형을 맞춰 주는 경우이다.

- `int + float`는 더 넓은 표현 범위를 가진 `float` 결과가 된다.
- `bool`은 숫자 연산에서 `True == 1`, `False == 0`처럼 처리된다.

```python
print('1 + 2.3 = ', 1 + 2.3)

print('True + 2 = ', True + 2)

print('True + True = ', True + True)

```

## 2. 명시적 형변환

명시적 형변환은 개발자가 `str()`, `int()`, `float()` 같은 함수를 사용해 직접 자료형을 바꾸는 방식이다.

- 문자열과 숫자는 바로 더할 수 없으므로 숫자를 문자열로 바꿔야 한다.
- `int()`로 실수를 정수로 바꾸면 소수점 아래는 버려진다.
- 숫자로 된 문자열은 `float()` 또는 `int()`로 변환해 계산에 사용할 수 있다.

```python
# print('안녕 + 123', '안녕'+123)
print('안녕 + 123', '안녕'+str(123))

height = 175.678
print('int:',int(height)) #소수점 버림
print('float:',float(height)) #정수 -> 실수 변환
print('float:', round(height))

print('--- 문자열 parsing ---')
str_number = ('123456')
print(str_number, type(str_number))
print(int(str_number), type(int(str_number)))

# print(int('일이삼'))
# #ValueError: invalid literal for int() with base 10: '일이삼'

```

## 3. 논리형으로의 형변환

조건식에서는 값이 `bool`로 변환되어 평가된다.

- **truthy**: 조건식에서 `True`처럼 평가되는 값
- **falsy**: 조건식에서 `False`처럼 평가되는 값
- 일반적으로 값이 있으면 `True`, 값이 비어 있거나 `0`이면 `False`로 평가된다.
- `''`는 빈 문자열이라 `False`이고, `' '`는 공백 문자 1개가 들어 있으므로 `True`이다.

```python
print('--- bool 형변환 : True ---')
print(bool(1)) #1을 bool 타입으로 형변환 시 True ==1 은 Truthy
print(bool(100))
print(bool(-2)) #0을 제외한 모든 정수는 Truthy
print(bool('hi')) #문자열도 True
print(bool(' ')) #문자열 공백도 True
print(bool([1, 2, 3])) # list, tuple, set, dict 등 값 1개라도 있으면 Truthy
print(bool((1, 2, 3)))
print(bool({1, 2, 3}))
print(bool({'price':1}))


# 숫자 0 또는 값 없는 경우 False
```

```python
print('--- bool 형변환 : False ---')
print(bool(0)) #0을 bool 타입으로 형변환 시 False==0 은 Falsy
print(bool('')) #문자열 값 없는 경우는 False

print(bool([])) # list, set, dict, tuple 값 없으면 Falsy
print(bool(()))
print(bool({}))

print(bool(None))

# 숫자 0 또는 값 없는 경우 False
```

## 4. 조건문에서의 활용

리스트, 튜플, 딕셔너리 같은 컬렉션은 요소가 있으면 `True`, 비어 있으면 `False`로 평가된다. 그래서 `len(lst) > 0`처럼 직접 길이를 비교하지 않아도 조건문에서 간단히 사용할 수 있다.

```python
lst = [1,2] # Truthy
if lst:
    print('lst에 값이 존재합니다')
else:
    print('lst에 값이 존재하지 않습니다')

```
