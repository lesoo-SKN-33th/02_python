## 파일 쓰기

```python
import os

f = open('helloworld2.txt', 'w', encoding='utf-8')
print(f, type(f))

text = '''
옛날치킨
생맥주
오왜월
오늘 왜 월요일이냐는 뜻
내일은 화요릴
'''

f.write(text) # 지정된 파일에 출력
# 기존 내용이 있으면 덮어쓰기
# 지정된 파일 존재하지 않으면 파일 생성


# 파일 연결에 사용된 메모리 반환
f.close()

```

```python
# with 절
# with 옆에 작성된 open() 구문에서 사용한 file 연결을 with 절 종료시 자동으로 close
with open('helloworld2.txt', 'w', encoding='utf-8') as f:
    f.write('with절을 이용하면 자동 close() 된다')
```

## 파일 읽기

```python
with open('helloworld.txt', 'r', encoding='utf-8') as f:
    print(f.read()) # 전부 읽어와서 출력
```

```python
#한줄씩 읽기
with open('helloworld.txt', 'r', encoding='utf-8') as f:
    for idx, line in enumerate(f):       #f.readlines():
        print(idx, line, end='')
```

```python
#파일 이어쓰기('a', append  모드)
with open('helloworld.txt', 'a', encoding='utf-8') as f:
    f.write(text)

with open('helloworld.txt', 'r', encoding='utf-8') as f:
    print(f.read())
```

```python
# 파일 복사
# 1. 파일읽기
with open('helloworld.txt', 'r', encoding='utf-8') as f:
    text = f.read() # with 절에 선언한 변수는 지역변수 x -> 전역 사용 가능
# 2. 파일 쓰기
with open('helloworld_copy.txt', 'w', encoding='utf-8') as f:
    f.write(text)
    print('파일복사 완료')
```

```python
# 이미지 파일 복사(아무 파일이나 가능)
# 1. 파일읽기
# - r,w,a 는 `문자`를 읽,쓰, 이어쓰기 모드
# - rb, wb 는 byte 단위로 읽,쓰기 모드

    # UnicodeDecodeError
    #이미지는 문자가 아니라서 encoding 하면 에러 발생
with open('princess.jpg', 'rb') as f:
    image_bytes = f.read()
# 2. 파일 쓰기
with open('copy_image.jpeg', 'wb') as f:
    f.write(image_bytes)
    print('파일복사 완료')
```

## dict-csv 읽고 쓰기

```python
# csv 읽기
# csv.DictReader(f)
# csv.DictWriter(f, fieldnames)
import csv
with open('users.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f) # 클래스임 -> dictReader 객체를 생성

    # for row in reader:
    #     print(row, type(row))
    #     users.append(row)
    #
    users = [row for row in reader]

    print(users)
```

```python
print(users)
```

```python
# csv 쓰기

with open('users.csv', 'w', encoding='utf-8', newline='') as f:
    # csv 쓰기 객체 생성
    # f: 쓸파일
    # fieldNames : 첫줄에 작성될 헤더(컬럼명)
    writer = csv.DictWriter(f, fieldnames=['id', 'name', 'email'])
    writer.writeheader()

    #users에 사용자 추가
    users.append({'id': 400, 'name': '이순신', 'email': 'sunshinlee@gmail.com'})
    users.append({'id': 500, 'name': '강감찬', 'email': 'gck@gmail.com'})
    print(users)


    #여러줄 쓰기
    # writer.writerows(users)
    #한줄씩 쓰기
    for user in users:
        writer.writerow(user)

```
