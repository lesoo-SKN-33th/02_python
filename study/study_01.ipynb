### Chapter 02. 자료형
p.152-153 도전문제
1. 구의 부피와 겉넓이 구하기

```python
r = int(input('구의 반지름을 입력해주세요'))
pi = 3.141592
fill = 4 / 3 * pi * (r ** 3)
out = 4 * pi * (r ** 2)
print(f'구의 부피는 {fill}입니다.')
print(f'구의 겉넓이는 {out}입니다.')
```

```python
a = float(input('밑변의 길이를 입력해주세요 : '))
b = float(input('높이의 길이를 입력해주세요 : '))
c = ((a ** 2) + (b ** 2)) ** 0.5

print(f'=빗변의 길이는 {c}입니다.')
```

### Chapter 03 조건문
p.188-189 도전문제
1. 간단한 대화 프로그램

```python
from datetime import datetime

now = datetime.now()
hour = now.strftime('%H')
print(hour)
while True:
    in_msg = input('입력 : ')
    if '안녕' in in_msg:
        print('안녕하세요.')
    elif '몇 시' in in_msg:
        print(f'지금은 {hour}시 입니다.')
    elif '종료' in in_msg :
        break
    else :
        print(in_msg)
```

2. 나누어 떨어지는 숫자

```python
number = int(input('정수를 입력해주세요 : '))
num_lst = [2,3,4,5]
for i in num_lst:
    print(f'{number}은 {i}로 나누어 떨어지는 숫자{'입니다' if number%i==0 else '가 아닙니다'}')

```

### Chapter 04 반복문
p.268-269 도전문제
1. 숫자의 종류

```python
lst = [1,2,3,4,1,2,3,1,4,1,2,3]
count = {}
for num in lst:
    if num not in count.keys():
        count[num] = 1
    else :
        count[num] += 1
st = set(lst)
print(f'{lst}에서 ')
print(f'사용된 숫자의 종류는 {len(st)}개 입니다.')
print(f'참고{count}')
```

2. 염기의 개수

```python
# dna = 'ctacaatgtcagtatacccattgcattagccgg'
dna = input('염기서열을 입력해주세요 : ')
count = {}
for c in dna:
    if c not in count.keys():
        count[c] = 1
    else :
        count[c] += 1
for key in count.keys():
    print(f'{key}의 개수 : {count[key]}')

```

3. 염기 코돈 개수

```python
# dna = 'ctacaatgtcagtatacccattgcattagccgg'
dna = 'ctacaatgtcagtatacccattgcattagccgg'
count = {}
codons = ['cta', 'caa', 'tgt', 'cag', 'tat', 'acc', 'cat', 'tgc', 'att', 'agc', 'cgg']
for codon in codons :
    for idx in range(0, len(dna)-2, 3): ##step 을 3으로 해야했음
        if dna[idx:idx+3] == codon:
            if codon not in count.keys():
                count[codon] = 1
            else :
                count[codon] += 1
print(count)


```

4. 2차원 리스트 평탄화

```python
lst = [1,2,[3,4], 5, [6,7], [8,9]]
new_lst = []
for i in range(0, len(lst)):
    if type(lst[i]) == list:
        for j in lst[i]:
            new_lst.append(j)
    else:
        new_lst.append(lst[i])
print(new_lst)
```
