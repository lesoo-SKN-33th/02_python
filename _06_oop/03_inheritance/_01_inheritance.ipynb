# 상속 Inheritance
- 부모클래스 속성/메서드를 자식클래스에서 선언없이 물려받아 사용하는 것
- 중복된 속성/메서드를 재작성할 필요가 없다.
- 자식클래스를 부모클래스로 묶어 관리할 수 있다.
- 파이썬 다중상속 지원

```python

class Person:
    def __init__(self, name: str, phone: str) -> None:
        self.__name = name
        self.__phone = phone

    def hello(self):
        print(f'안녕하세요, 저는 {self.__name}입니다. 제 연락처는 {self.__phone}이에요.')

    def get_name(self):
        return self.__name

    def get_phone(self):
        return self.__phone

    # __str__(): print(객체)를 했을 때 어떤 문자열을 보여줄지 정하는 메서드
    def __str__(self):
        return f'Person(name={self.__name}, phone={self.__phone})'



p1 = Person('홍길동', '01012341234')
print(p1) # __str 메소드 없으면 메모리 주소 나옴




```

```python
#Person 상속받은 자식 클래스
class Employees(Person):
    def __init__(self, name: str, phone: str, company:str) -> None:
        # self.__name = name -> 사용 불가 : person 객체의 속성이라 직접 수정 불가

        #부모 생성자 super()
        super().__init__(name, phone)
        # 자식이 따로 세팅
        self.__company = company

    def hello(self):
        #method overriding
        # 부모로부터 상속받은 hello 메서드를 재작성(재정의
        super().hello() # 부모 메서드 호출
        print(f'{self.__company}에서 근무중')

    def __str__(self):
        return f'Employee(name={self.get_name()}, phone={self.get_phone()}, company={self.__company})'

emp1 = Employees('홍길동', '01021341234', 'SKN')
print(emp1.get_phone()) # 상속받은 메소드 호출
print(emp1.get_name())
emp1.hello()
# emp1.get_company()
print(emp1)
```

```python
class Developer(Person):

    # *args == *langs == 3번째 이후 매개변수를 모두 모아 tuple로 저장
    def __init__(self, name: str, phone: str, *langs:tuple[str]) -> None:
        super().__init__(name, phone)
        self.__langs = list(langs) #tuple -> list 변환
    #langs에 언어 추가
    # 언어(lang)을 전달 받아와서 langs에 없으면 추가
    # langs에 이미 존재하는 언어면 '이미 추가되어있습니다' 출력
    def add_lang(self, lang:str):
        if lang not in self.__langs:
            self.__langs.append(lang)
            print(f'{lang} 추가되었습니다.')
        else:
            print(f'{lang}은 이미 추가되어 있습니다.')

    #langs에 언어 제거
    # 언어(lang)을 전달 받아와서 langs에 있으면 제거
    # langs에 이미 존재하지 않는 언어면 '존재하지 않는 언어입니다' 출력
    def remove_lang(self, lang:str):
        if lang in self.__langs:
            self.__langs.remove(lang)
            print(f'{lang} 제거되었습니다.')
        else:
            print(f'{lang}은 존재하지 않는 언어입니다')

    # hello() overriding
    # 먼저 부모의 hello 출력 후 '사용가능한 언어 목록 출력
    # '사용가능한 언어 목록 : ',{langs}
    def hello(self):
        super().hello()
        print(f'사용가능한 언어 목록 : {self.__langs}')

    #__str__() overring
    # developer(name={}, phone={}, langs={}) 출력
    def __str__(self):
        return f'developer(name={self.get_name()}, phone={self.get_phone()}, langs={self.__langs})'


dev1 = Developer('홍길동', '01012341234', 'kor', 'jpn', 'chn', 'eng')

print(dev1)
dev1.add_lang('eng')
dev1.remove_lang('eng')
print(dev1)
dev1.add_lang('eng')
print(dev1)
dev1.hello()
```

## isinstance(), issubclass(), MRO

상속 관계에서는 객체가 어떤 클래스의 인스턴스인지, 클래스가 어떤 부모를 상속하는지 확인할 수 있다.

- `isinstance(객체, 클래스)`: 객체가 해당 클래스 또는 하위 클래스의 인스턴스인지 확인
- `issubclass(자식클래스, 부모클래스)`: 클래스 사이의 상속 관계 확인
- MRO(Method Resolution Order): 메서드를 찾는 순서


```python
# dev1 변수가 Developer 클래스로 만든 인스턴스 맞는지 확인
print('isinstance(dev1, Developer)', isinstance(dev1, Developer))
print('isinstance(dev1, Person)', isinstance(dev1, Person))
print('isinstance(dev1, Employees)', isinstance(dev1, Employees)) #객체끼리 형제관계는 남임

print('issubclass(Developer, Person)', issubclass(Developer, Person))
print('issubclass(Person, Developer)', issubclass(Person, Developer))

print('MRO(Developer):', Developer.mro())

#object 클래스 == 모든 클래스의 부모
# class Test: == class Test(object):
# 상속 부문이 없는 클래스는 object 상속이 자동 추가된다



```

## Python의 오버라이딩과 오버로딩 주의

Python은 자식 클래스에서 부모 클래스의 메서드를 다시 정의하는 오버라이딩은 자연스럽게 지원한다.
다만 Java처럼 같은 이름의 메서드를 매개변수 조합별로 여러 개 선언하는 메서드 오버로딩은 지원하지 않는다.

같은 클래스 안에서 같은 이름의 메서드를 다시 선언하면, 뒤에 선언한 메서드가 앞의 메서드를 덮어쓴다.


```python
class Calculator:
    def add(self, a): # 얘는 동작 안함. 같은 이름이면 나중에 있는애가 덮어씀
        return a

    def add(self, a, b):
        return a + b


calculator = Calculator()

# calculator.add(10) # b인자 빠졌다고 에러 내용
# TypeError: Calculator.add() missing 1 required positional argument: 'b'
#같은 이름 메서드 중 나중에 선언된 메서드만 사용 가능

calculator.add(10, 20)

```
