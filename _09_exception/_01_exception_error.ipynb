## 예외처리
**(중요) 발생된 예외/오류 상황에서 try..except구문을 통해 적절히 처리후 정상적인 흐름을 돌려내는 것**

```python
print('시작')
# try : 예외가 발생할 가능성이 있는 코드를 작성하는 구문
try:
    result = 10/0 # 예외가 발생할수 있는 코드

    # 예외 발생 원리
    # 1. 예외 상황 감지
    # 2. 해당 예외 상황에 맞는 '예뢰 클래스' 찾기
    # 3. 예외클래스를 이용하여 '예외 객체'생성
    # 4. 생성된 예외 객체를 예외 상황이 감지된 코드위치로 올림(raise)


except ZeroDivisionError as e:
    # except : try에서 올려진 예외 처리 구문
    print(e)
    print('0으로 나눌수 없습니다')

    # except 구문이 종료되면 다시 정상흐름으로 복귀

print('종료')
```

```python
# except 구문을 이용한 예외 객체 사용법
import random # 난수 생성


try:
    num = random.randint(0, 1) # 0-1사이 정수형 난수 생성 -> 0or 1 생성됨
    print(num)

    # 0 == False, 1 == True
    if num: # 1인경우
        print('abc'+123)# TypeError
    else: #0인경우
        print(int('abc')) # ValueError
except TypeError as e:
    print(e)
    print('예외타입 이름: ', type(e).__name__)
    print('예외 메세지: ', str(e))
    print('예외 상세값: ', e.args)
    print(f'[Error]{type(e).__name__} : {str(e)}')
except ValueError as e:
    print(e)
    print('예외타입 이름: ', type(e).__name__)
    print('예외 메세지: ', str(e))
    print('예외 상세값: ', e.args)
    print(f'[Error]{type(e).__name__} : {str(e)}')
    #e는 try에서 발생한 예외객체

```

```python
# 숫자를 입력받아 입력받은 숫자와 key가 일치하는 학생 조회하여 id, name 출력
# 입력받은 숫자 없으면 1-3 사이 숫자만 입력하라고 설정
# 잘못된 숫자 입력 시 다시 숫자 입력 받기

students: dict[int, dict] = {
    1: {"id": 10, "name": "홍길동"},
    2: {"id": 20, "name": "신사임당"},
    3: {"id": 30, "name": "이순신"},
}
while True:
    try:
        num = int(input('숫자 입력'))
        print(num)
        student:dict = students[num]
        print('id: ', student['id'])
        print('name: ', student['name'])
        break
    except ValueError as e:
        print(e)
    except KeyError as e:
        print(e)
        print('1-3 사이의 숫자만 입력하세요')

```

### pass
- if, for, def, class, except 등의 블록은 내부가 비어있으면 오류가 발생함
- 이 때, 구조만 유지하는 용도 또는 예외를 무시할 때 사용

```python
if True:
    pass

def func():
    pass
```

```python
# 예외 무시
try:
    print(10/0)
except :
    pass
```

### finally
- try 구문에서 예외 발생 여부와 관계없이 마지막에 무조건 실행하는 구문

### else
- try 구문에서 예외가 발생되지 않았을때 수행되는 구문

```python
try:
    num = int(input('숫자입력'))
    result = 100/num
    # print(result)
except ZeroDivisionError as e:
    print(e, '0으로 나눌수 없음')
except ValueError as e:
    print(e, '정수만 입력')
else:
    print(result)
finally:
    print('종료')
```

```python
# finally 사용 예시
f = open('helloworld.txt', 'r', encoding='utf-8')
try:
    print(f.read())
    print(10/0)
except :
    pass
finally:
    f.close()
```
