## map
iterable객체의 각요소에 대해 지정한 함수를 적용하고, 그 결과를 map객체로 반환
```
map(function, iterable) -> map
```




````python
nums:list = [1,2,3,4,5]

result = map(lambda x: x**2, nums)
print(result, type(result))
print('list(result):', list(result))

# map : 기존 list의 요소를 하나씩 꺼내 함수를 호출
# 함수 결과들을 모아 새로운 list 생성

````

````python
# 문자열 리스트 -> 문자열 길이 리스트

products:list[str] = ['아아메', '아바라', '수박주스', '말차라떼', '식혜']

product_len:list[int] = map(lambda x: len(x), products)#(콜백함수, 순회가능 객체)
print(product_len, list(product_len), type(product_len))

````

````python
# 여러 순회 가능 객체(list)통합

mid_term = [90, 55, 70, 78]
final_term = [87,70, 70, 95]

avg_score = list(map(lambda x,y:(x+y)/2, mid_term, final_term))
print(avg_score)

````

## filter

주어진 iterable객체에 대해서 참/거짓 평가후 참인 요소들만 결과값으로 반환

```
filter(function, iterable) -> filter
```

````python
nums = list(range(1, 11))
print(nums)

evens = filter(lambda n:n%2==0, nums)
print(list(evens), type(evens))

odds = list(filter(lambda n:n%2==1, nums))
print(odds)

threes = list(filter(lambda n:n%3==0, nums))
print(threes)
````

````python
# 이름 목록에서 성이 홍씨인 사람 찾아서 리스트로 반환
names:list[str] = ['홍길동', '고길동', '홍진경', '최홍만']
hongs = list(filter(lambda str:str[0]=='홍', names))
hong_family = list(filter(lambda name:name.startswith('홍'), names))
# str.startswith(str2) -> bool
# 문자열의 시작이 str2와 같다면 True

print(hongs)
print(hong_family)

````

````python
# dict 검색(dict 필터링)
# 학생 목록에서 점수가 90 이상인 학생만 필터링)
students = [
    { 'name' : '이순신', 'score' : 80},
    { 'name' : '윤봉길', 'score' : 75},
    { 'name' : '논개', 'score' : 99},
    { 'name' : '강감찬', 'score' : 91},
]

student1 = list(filter(lambda x : x['score'] >= 90, students))
print(student1)

student2 = [s for s in students if s['score'] >= 90]
print(student2)

````

## reduce
iterable객체의 요소에 대한 연산결과를 단일값을 반환.
합계, 최대/최소값 등의 연산에 적합

```
reduce(function, iterable, initial_value) -> Any
```
함수는 3개의 인자를 받는다.
1. 누적 연산값
2. 현재 요소
3. 초기값 (선택)


````python
from functools import reduce # 외부 functools 모듈에서 reduce 함수만 가져오기

# 정수형 list 내 요소들의 합계
nums:list[int] = [11,22,33,44,55,66,77,88,99]
total = reduce(lambda sum,curr:sum+curr, nums, 0)
total2 = reduce(lambda x,y:x+y, nums, 5)# 시작값 5로 해놔서 +5 되고서 시작함
#reduce(콜백함수(누적용 변수(x), 현재값(y)), 순회객체, 누적변수 초기값)
# -> 콜백함수의 return 값이 자동으로 누적용 변수에 대입된다
print(total)
print(total2)

````

````python
# 최대/최소값 찾기
nums : list[int] = [23,34,67,76,32,78,89,90,37,58,-1]
max_num = reduce(lambda max, curr: max if max>curr else curr, nums, nums[0])
min_num = reduce(lambda min, curr: min if min<curr else curr, nums, nums[0])
print(max_num)
print(min_num)

# 절대값이 가장 큰 수
def abs_func(max, curr):
    print(f'max:{max}, curr:{curr}')
    return max if abs(max) > abs(curr) else curr


abs_max = reduce(abs_func, nums, 0)
print(abs_max)
````

````python
students = [
    { 'name' : '이순신', 'score' : 80},
    { 'name' : '윤봉길', 'score' : 75},
    { 'name' : '논개', 'score' : 99},
    { 'name' : '강감찬', 'score' : 91},
]

best_score = reduce(lambda x,c : x if x['score'] > c['score'] else c, students)
print(best_score)
````
